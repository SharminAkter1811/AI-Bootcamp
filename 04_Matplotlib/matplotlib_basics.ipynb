```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data={
    "Name":["akash","Nibir","Payel","Trina","Joya"],
    "Age":[23,25,24,26,22],
    "Score":[89,78,65,50,90]
}
df=pd.DataFrame(data)
df
```

```python
names = ["Asha", "Rafi", "Nila", "Tanvir", "Mim"]
scores = [85, 72, 91, 68, 88]
```

```python
plt.bar(names,scores)
plt.xlabel("Students")
plt.ylabel("Scores")
plt.title("Student scores")
plt.show()
```

Line graph


```python
plt.plot(names,scores,marker="o")
plt.xlabel("Students")
plt.ylabel("Scores")
plt.title("Student scores")
plt.show()
```

Scatter plot


```python
ages = [22, 25, 21, 28, 24]
scores = [85, 72, 91, 68, 88]
plt.scatter(ages,scores)
plt.xlabel("Ages")
plt.ylabel("Scores")
plt.title("Age")
plt.show()
```
