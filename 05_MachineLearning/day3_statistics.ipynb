```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data={
    "Study_Hours" : [10,10,5,2,8,9,3,1,5,6,4,7],
    "Sleep_Hours" : [8,6,8,8,7,7,4,6,5,9,8,7],
    "Attendance" : [90,100,60,80,40,79,80,60,70,50,90,75],
    "Score" : [98,96,80,40,70,89,50,47,65,78,95,57]
}

df=pd.DataFrame(data)
df
```

```python
print(df["Score"].mean())
print(df["Score"].median())
print(df["Score"].std())
print(df["Score"].max())
print(df["Score"].min())
```

```python
corr1=df["Study_Hours"].corr(df["Score"])
print(corr1)

corr2=df["Sleep_Hours"].corr(df["Score"])
print(corr2)

corr3=df["Attendance"].corr(df["Score"])
print(corr3)
```

```python
plt.scatter(df["Study_Hours"],df["Score"])
plt.xlabel("Study_Hours")
plt.ylabel("Score")
plt.title("Study_hours VS Score")
plt.show()
```

```python
plt.scatter(df["Sleep_Hours"],df["Score"])
plt.xlabel("Sleep_Hours")
plt.ylabel("Score")
plt.title("Sleep_hours VS Score")
plt.show()
```

```python
plt.scatter(df["Attendance"],df["Score"])
plt.xlabel("Attendance")
plt.ylabel("Score")
plt.title("Attendance VS Score")
plt.show()
```
